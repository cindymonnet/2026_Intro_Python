## Exercise 7: while, break, and continue

A logger returned six readings, two of which are missing:

```python

```

Using a `while` loop over the indices:

1. Skip any missing value with `continue`.
2. Stop at the first reading above 9.0 °C with `break`, storing it in `first_hot`.
3. Print `first_hot` after the loop. Define it before the loop so the code is safe even if no
   reading exceeds the threshold.

````python
mixed_celsius = [6.1, None, 5.9, None, 9.5, 6.8]
i = 0                                              # position de départ de l'incrément
first_hot = None                                   # définition pour que le code soit safe

while i < len(mixed_celsius):                      # pendant que i est en-dessous du nombre de position dans mixed_celsius
  value = mixed_celsius[i]                         # stocker la valeur pour les positions
  i += 1                                           # update de la position
  print("Température:", value)                     # print de la valeur assignée à la position
  if value == None:                                # si la valeur à la position est None
    continue
  if value > 9.0:
    first_hot = value                              # stocker la valeur correspondant au test dans first_hot
    break

print("First hot day:", first_hot)                 # print la valeur stockée


# Hint: continue jumps straight back to the while line, skipping everything below it in the
# loop body, including an index update at the bottom of the loop
````

returns **two** things: the total and the mean.

   `total_celsius` and `mean_celsius`.
2. Call it again with `scale=2.0` passed by keyword, and print only the mean.

````python
temperatures = [6.1, 7.4, 5.9, 8.2, 6.8]        # nom de la liste à l'extérieur de la boucle
def summarise(values, scale=1.0):               # création de la liste avec un nom générique à l'intérieur de la boucle
  total = 0
  for v in values:                              # nom générique de la liste dans la boucle
    total += v * scale
  return total, total/len(values)               # toujours le nom à l'intérieur de la boucle

total_celsius, mean_celsius = summarise(temperatures, scale=1.0)        # dis à Python quelle liste choisir
print(total_celsius, mean_celsius)

total_celsius, mean_celsius = summarise(temperatures, scale=2.0)
print(mean_celsius)

result = summarise(temperatures, scale=2.0)
print(result, type(result))                     # dans la boucle, Python comprend values = temperatures


# PEUT ÊTRE UTILISE COMME DIFFICULT EXERCISE
````

## Exercise 9: A variable number of arguments

Write a function `warmest(*temps_celsius)` (no type hints) that returns the highest value among
any number of positional arguments, using a loop and a running maximum (not `max`). Test it on
three individual readings passed directly, and again by unpacking the list
`[6.1, 7.4, 5.9, 8.2, 6.8]` with `*`.

````python
temperatures = [6.1, 7.4, 5.9, 8.2, 6.8]        # création de la liste
def warmest(*temps_celsius):                    # def -> il existe une fonction appelée warmest qui accepte n'importe quel nombre d'arguments (*), tempcelsius regroupera les arguments dans ce tupple
  maximum = temps_celsius[0]                    # record de départ (première valeur lue)
  for t in temps_celsius:                       # test les valeurs du tupple dans chaque tour
    if maximum < t:                             # condition à respecter
      maximum = t                               # remplace le maximum par la valeur lue si elle respecte la condition
  return maximum                                # renvoie de la valeur maximale

print(warmest(2, 15, 3))
print(warmest(*temperatures))                   # l'étoile éparpille la liste en arguments séparés
````

### Exercise 14: Spot the impure function

This function is meant to return the readings without the last one:

```python
def drop_last(readings):
    readings.pop()
    return readings
```

1. Call it on `original = [6.1, 7.4, 5.9]`, then print `original`. What happened?
2. Rewrite it as a pure function that leaves the caller's list untouched, and demonstrate that
   `original` is unchanged after the call.
3. In one comment, connect this to the mutable-default bug from the lecture.

````python
def drop_last(readings):
    readings.pop()            # remove the last item from the list original list instead of giving a shorter list -> inpure function
    return readings

original = [6.1, 7.4, 5.9]
result = drop_last(original)
print(original)               # the last value is gone, here the original list is modified because readings and original are 2 names for the same list
````

````python
def drop_last(readings):
    return readings[:-1]      # removes the value from a new list and not the original -> pure function

original = [6.1, 7.4, 5.9]
result = drop_last(original)
print(original)               # the original is not changed, because we have 2 different lists
print(result)
````

### Exercise 15: Reject impossible temperatures

Write `to_kelvin(temp_celsius)` that raises a `ValueError` when the input is below absolute zero
(−273.15 °C) and otherwise returns the temperature in kelvin. Give it a docstring and a type hint.

Loop over `[20.0, -300.0, 0.0]`, calling it inside a `try`/`except` so that the invalid value is
reported without stopping the loop.

````python
def to_kelvin(temp_celsius: float) -> float:        # we say what temp_celsius must be and what we want the function to give
  """ Convert °C in °K and give a ValueError if the temperature in °C is below absolute 0. """
  if temp_celsius < -273.15:
    raise ValueError(f'{temp_celsius}')
  return temp_celsius + 273.15

temperatures = [20.0, -300.0, 0.0]

for t in temperatures:
  try:
    kelvin = to_kelvin(t)
    print(f'{t} in °C = {kelvin} in °K')
  except ValueError as error:
    print(f'{error}°C is below the absolute 0 (-273.15°C)')
````
