## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm). Print each result to two decimals with an f-string.

````python
temp = 18.5                     # temperature in degrees Celsius
temp_kelvin = temp + 273.15     # temperature in degrees Kelvin
temp_fahr = temp * 1.8 + 32     # temperature in degrees Fahrenheit
catchement_area = 250           # catchement area in km2
print(catchement_area)

# Q = v*A                       # debit equation
amount_oneday = 1*60*60*24      # m3 per day
print(amount_oneday)
dis_oneday_area = amount_oneday / 1000000 # m3 per day on a 1 km2 surface
print(dis_oneday_area)
v = dis_oneday_area*1000             # one day discharge in mm
print(v)

# from debit equation Q = v*A, where Q is the discharge and A the catchement area, we look for the speed = debit
spec_discharge = discharge/catchement_area
print(spec_discharge)
final_dis = spec_discharge*v

print(f"Discharge of {final_dis:.2f} mm/day")
print(f"{temp}°C {temp_kelvin:.2f}°K {temp_fahr:.2f}°F") # always put f at the very beginning, then open " and each variable between brackets with the f treatment but the text we want to print outside the brackets


## COULD BE CHOSEN AT CHALLENGING
````

## Exercise 3: Operator families and a quality flag

Three daily readings arrive from a station: `reading_1_celsius` = 18.2, `reading_2_celsius` = -300.0, and `reading_3_celsius` = None (the sensor reported nothing).

1. Build a boolean `is_plausible` that is `True` when the first reading is above absolute zero and below 60 °C.
2. Build `is_implausible` for the second reading: apply the same test as in step 1, and negate it with `not`.
3. Use `is` to check whether the third reading is missing.
4. Starting from `n_valid = 0`, use `+=` twice to count how many of the first two readings pass the step 1 test. Note that `int()` casts `True` to 1 and `False` to 0.
5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use `>=` and `<=` to test whether the first reading lies in that range, and `!=` to check that the first two readings differ.

````python
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None
is_plausible = -273.15 < reading_1_celsius and reading_1_celsius < 60
print(is_plausible)

is_implausible = not (-273.15 < reading_2_celsius and reading_2_celsius < 60)
print(is_implausible)

print(reading_3_celsius is None)

n_valid = 0
n_valid += int(-273.15 < reading_1_celsius and reading_1_celsius < 60)
n_valid += int(-273.15 < reading_2_celsius and reading_2_celsius < 60)
print(n_valid)

reading_1_celsius >= -40
reading_1_celsius <= 60
reading_1_celsius != reading_2_celsius
````

## Exercise 7: Round-trip through a file, with an append

Write the Jungfraujoch week from the lecture — the daily maximum temperatures -8.4, -8.9, -9.2,
-9.0, -9.5 (°C) — to `_files/jfj_week.txt`, one value per line, using `pathlib` and a `with`
block. Create the `_files` folder first if it does not exist; do not rely on Exercise 6 having
made it.

1. Write the five values with mode `"w"`, one `.write()` call per value.
2. Reopen the file with mode `"a"` and append the last two days of the week, `-7.8` and `-6.1`.
3. Read the file back with `.readlines()`, cast each of the seven lines to `float` by name, and
   compute the mean by summing the seven values by hand with `+` and dividing by `len()`. Print
   it to two decimals.

````python
from pathlib import Path

data_path2 = Path("_files/jfj_week.txt")
data_path2.parent.mkdir(exist_ok=True)
print(data_path2)

with data_path2.open("w", encoding = "utf=8") as fhandler:    # utf=8 dit comment lire ou écrire les caractères -> standard universel
  fhandler.write("-8.4°C\n")
  fhandler.write("-8.9°C\n")
  fhandler.write("-9.2°C\n")
  fhandler.write("-9.0°C\n")
  fhandler.write("-9.5°C\n")

with data_path2.open("a", encoding = "utf=8") as fhandler:
  fhandler.write("-7.8°C\n")
  fhandler.write("-6.1°C\n")

with data_path2.open("r", encoding = "utf=8") as fhandler:
  raw_text2 = fhandler.readlines()

print(raw_text2)

temp1 = float(raw_text2[0].replace("°C", ""))
temp2 = float(raw_text2[1].replace("°C", ""))
temp3 = float(raw_text2[2].replace("°C", ""))
temp4 = float(raw_text2[3].replace("°C", ""))
temp5 = float(raw_text2[4].replace("°C", ""))
temp6 = float(raw_text2[5].replace("°C", ""))
temp7 = float(raw_text2[6].replace("°C", ""))

print(temp1, temp2, temp3, temp4, temp5, temp6, temp7)

total = (temp1 + temp2 + temp3 + temp4 + temp5 + temp6 + temp7)/len(raw_text2)     # len is to divide the string into the number of elements it contains
print(f"{total:.2f}")
````
